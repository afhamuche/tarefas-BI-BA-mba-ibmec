```python
%sql
SELECT
COUNT(*)          
AS pedidos,
MIN(o_orderdate)  AS primeira_data,
MAX(o_orderdate)  AS ultima_data
FROM samples.tpch.orders;
```

```python
%sql
SELECT
date_trunc('MONTH', o_orderdate) AS mes,
SUM(o_totalprice)                
AS receita,
COUNT(*)                         
FROM samples.tpch.orders
GROUP BY 1
ORDER BY 1;
```

```python
%sql
SELECT
date_trunc('MONTH', o_orderdate)  AS mes,
COUNT(DISTINCT o_orderdate)       
AS dias_com_pedido,
SUM(o_totalprice)                 
AS receita,
SUM(o_totalprice) / COUNT(DISTINCT o_orderdate) AS receita_por_dia
FROM samples.tpch.orders
WHERE o_orderdate >= DATE '1998-08-02'
GROUP BY 1
ORDER BY 1;
```

```python
%sql
SELECT
c.c_mktsegment  AS segmento,        
SUM(o.o_totalprice)     AS receita,
COUNT(*)                AS pedidos,
AVG(o.o_totalprice)     AS ticket_medio




FROM samples.tpch.orders   AS o
JOIN samples.tpch.customer AS c
ON o.o_custkey = c.c_custkey
GROUP BY 1
ORDER BY receita DESC;
```

```python
%sql
SELECT
ROUND(AVG(o_totalprice), 2)                     AS media,
ROUND(percentile_approx(o_totalprice, 0.5), 2)  AS mediana,
ROUND(MIN(o_totalprice), 2)            AS minimo,         
ROUND(MAX(o_totalprice), 2)                   AS maximo  
FROM samples.tpch.orders;
```

```python
%sql
SELECT
FLOOR(o_totalprice / 25000) * 25000  AS faixa_valor,
COUNT(*)     AS pedidos
                        
FROM samples.tpch.orders
GROUP BY 1
ORDER BY 1;
```
